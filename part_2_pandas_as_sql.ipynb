# 📊 Part 2: Working with Data (Pandas as SQL)

Now that we know the basic building blocks, let's introduce **Pandas**.
Pandas is the standard data manipulation library in Python. It allows us to work with data in a tabular format (rows and columns), just like a SQL database.

In this notebook, we will translate your SQL vocabulary directly into Python Pandas syntax!

```python
import pandas as pd
import seaborn as sns

# Load a real public dataset: 'tips'
# This contains data about restaurant bills, tips, and customer demographics
df = sns.load_dataset('tips')

print("Table 'df' (Tips Dataset) loaded successfully!")
display(df.head())
```

### 1. `SELECT *` (Inspecting Data)
**SQL:** `SELECT * FROM tips LIMIT 5;`

**Pandas:** `.head()` and `.info()`

```python
# EXAMPLE: View the first 5 rows and check schema
display(df.head(5))
df.info()
```

### 2. Selecting Columns & Aggregates (`SELECT col`)
**SQL:** `SELECT total_bill, tip FROM tips;` | `SELECT MAX(total_bill) FROM tips;`

**Pandas:** `df[['col1', 'col2']]`

**Example:**

```python
# SELECT day, time FROM df LIMIT 3;
subset = df[['day', 'time']]
display(subset.head(3))

# SELECT MAX(total_bill) FROM df;
max_bill = df['total_bill'].max()
print(f"\nThe maximum bill amount is: ${max_bill}")
```

### 3. Filtering Data (`WHERE`)
**SQL:** `SELECT * FROM tips WHERE day = 'Sun' AND total_bill > 20;`

**Pandas:** `df[(df['col'] == 'val')]`

**Example:**

```python
# SELECT * FROM df WHERE size > 4 OR smoker = 'Yes';
filtered_df = df[(df['size'] > 4) | (df['smoker'] == 'Yes')]
display(filtered_df.head(100))
```

### 4. Aggregations & Sorting (`GROUP BY` & `ORDER BY`)
**SQL:** `SELECT time, SUM(total_bill) FROM tips GROUP BY time ORDER BY SUM(total_bill) DESC;`

**Example:**

```python
# Grouping, aggregating, and sorting
# We add observed=False to handle categorical data and remove the warning
revenue_by_time = df.groupby('time', observed=False)['total_bill'].sum().sort_values(ascending=False)
display(revenue_by_time)
```

### 5. Calculated Columns & Exporting
**SQL:** `SELECT *, (tip/total_bill)*100 AS tip_percentage FROM tips;`

**Pandas:** `df['new_col'] = ...`

**Example:**

```python
# Creating a new computed column: Tip Percentage
df['tip_percentage'] = (df['tip'] / df['total_bill']) * 100

# Let's check the new column
display(df[['total_bill', 'tip', 'tip_percentage']].head())

# Exporting to CSV
df.to_csv('processed_tips.csv', index=False)
print("\nData successfully exported to 'processed_tips.csv'!")
```

### 6. Joining Data (`JOIN`)
**SQL:** `SELECT * FROM tips LEFT JOIN staff ON tips.day = staff.day;`

**Pandas:** `pd.merge(df1, df2, on='key', how='inner|left|right|outer')`

**Example (Different Join Types):**

```python
# 1. Create staff data
staff_data = pd.DataFrame({
    'day': ['Thur', 'Fri', 'Sat', 'Sun', 'Mon'], # Mon is extra
    'manager': ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
})

display(staff_data)

# 2. Left Join (Keep all rows from 'tips')
left_join = pd.merge(df, staff_data, on='day', how='left')

# 3. Outer Join (Keep everything from both, filling NaNs where no match)
outer_join = pd.merge(df, staff_data, on='day', how='outer')

print("Left Join result (First 5 rows):")
display(left_join.head())

print("\nCheck the unique managers in Outer Join (includes Eve):")
print(outer_join['manager'].unique())
```
