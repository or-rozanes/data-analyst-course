### 5. Simulating a JOIN
**Concept:** Combining two datasets based on a common key.

**SQL Connection:**
This is the equivalent of an `INNER JOIN`. We take a value from one table (like `city_id`) and use it to find the matching row in another table to get more details (like the `city_name`).

````python
# --- Example: Dictionary Lookup (JOIN behavior) ---
# Table A: Cities
cities = {
    1: "Tel Aviv",
    2: "New York",
    3: "London"
}

# Table B: Users with a 'city_id' column
users = [
    {"id": 101, "name": "Alice", "city_id": 1},
    {"id": 102, "name": "Bob", "city_id": 3}
]

# Performing the 'Join' in a loop
for user in users:
    city_name = cities.get(user['city_id'], "Unknown")
    print(f"{user['name']} lives in {city_name}")
````

**SQL Equivalent:**
```sql
SELECT
    u.name,
    c.city_name
FROM users u
JOIN cities c ON u.city_id = c.id;
```
